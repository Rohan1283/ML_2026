```python
# One hot encoding using pandas

pd.get_dummies(cr, columns=['fuel','owner'], dtype=int)
```

```python
# K-1 onehot encoding ( to avoid colinearity)


pd.get_dummies(cr, columns=['fuel','owner'], dtype=int, drop_first = True)
```

```python
# Drawback of using pandas for onehot encoding is pandas does not remembers the positioning of different columns.
# that is why use OneHotEncoder of sklearn

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(cr.iloc[:,:4], cr.iloc[:,-1], test_size = 0.2, random_state =0)
```

```python
# Ordinal encoding on cough

oe = OrdinalEncoder(categories = [['Mild','Strong']])
X_train_cough = oe.fit_transform(X_train[['cough']])

X_test_cough = oe.transform(X_test[['cough']])
X_train_cough.shape
```

```python
# OneHotEncoding of gender and city

ohe = OneHotEncoder(drop ='first', sparse_output = False)

X_train_gender_city = ohe.fit_transform(X_train[['gender','city']])

X_test_gender_city = ohe.transform(X_test[['gender', 'city']])

X_train_gender_city
```

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer(transformers =[
    ('tnf1', SimpleImputer(),['fever']),
    ('tnf2', OrdinalEncoder(categories = [['Mild','Strong']]), ['cough']),
    ('tnf3', OneHotEncoder(sparse_output = False, drop ='first'), ['gender','city'])
], remainder ='passthrough')
```

```python
transformer.fit_transform(X_train)
transformer.transform(X_test)
```
