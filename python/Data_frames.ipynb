```python
def euclidean_dist(point):
    point_1 = point['Ist_point'] 
    point_2 = point['IInd_point']

    return ((point_1[0] - point_2[0])**2 + (point_1[1] - point_2[1])**2)**0.5
```

```python
dest_pos = pd.DataFrame({
    'Ist_point' : [(0,3), (-5,2), (3,4)],
    'IInd_point' : [ (3,4), (3,6) , (-4,-4)]
})
```

```python
dest_pos['euclidean_distance'] = dest_pos.apply(euclidean_dist, axis =1)
```
