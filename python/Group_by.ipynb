```python
def fo(group):
    return group['Series_Title'].str.startswith('A').sum()
```

```python
genre.apply(fo)
```

```python
def rank_group(group):
    group['IMDB_ranking'] = group['IMDB_Rating'].rank(ascending = False, method = 'min')
    # can use method ='dense' or 'first' as well both has different properties.
    return group
```

```python
genre.apply(rank_group)
```

```python
def normal(group):
    group['norm_rating'] = (group['IMDB_Rating']-  group['IMDB_Rating'].min())/(group['IMDB_Rating'].max() - group['IMDB_Rating'].min())
    return group

genre.apply(normal)
```

```python
kohli_df = ipl[ipl['batsman'] == 'V Kohli'].copy()

# 2. Create boolean columns for fours and sixes (True = 1, False = 0)
kohli_df['fours'] = kohli_df['batsman_runs'] == 4
kohli_df['sixes'] = kohli_df['batsman_runs'] == 6

# 3. Group by team and sum up the runs, fours, and sixes
output = kohli_df.groupby('bowling_team').agg({
    'batsman_runs': 'sum',
    'fours': 'sum',
    'sixes': 'sum'
}).sort_values(by='batsman_runs', ascending=False).reset_index()
```

```python
output
```

```python
def highest_score(batsman):
    temp_df = ipl[ipl['batsman'] == batsman]
    return temp_df.groupby('match_id')['batsman_runs'].sum().sort_values(ascending = False).head(1).values[0]
```
