# Bivariate Analysis

```python
titanic = pd.read_csv('Titanic-Dataset.csv')
```

```python
iris = pd.read_csv('Iris.csv')
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
flights = sns.load_dataset('flights')
```

```python
flights
```

## 1. Scatterplot ( Numerical - Numerical)

```python
# Bivariate Analysis

sns.scatterplot(x ='total_bill', y= 'tip', data = tips)
```

```python
# MultiVariate Analysis

sns.scatterplot(x ='total_bill', y= 'tip', data = tips, hue = 'sex')
```

```python
sns.scatterplot(x ='total_bill', y= 'tip', data = tips, hue = 'sex', style = 'smoker')
```

```python
sns.scatterplot(x ='total_bill', y= 'tip', data = tips, hue = 'sex', style = 'smoker', size ='size')
```

## 2. Bar plot ( Numerical - Categorical)

```python
titanic.head()
```

```python
sns.barplot(x='Pclass', y = 'Age', data = titanic)
```

```python
sns.barplot(x='Pclass', y = 'Fare', data = titanic)
```

```python
sns.barplot(x='Pclass', y = 'Fare', data = titanic, hue = 'Sex')
```

## 3 Boxplot ( Numerical - Categorical)

```python
sns.boxplot(x= 'Sex' , y ='Age', data = titanic, hue ='Survived')
```

## 4. Distplot  ( Numerical - Categorical)

```python
sns.distplot(titanic[titanic['Survived']==0]['Age'],hist = False)
sns.distplot(titanic[titanic['Survived']==1]['Age'],hist = False)
```

## 5. HeatMap (Categorical - Categorical)

```python
titanic.head(3)
```

```python
sns.heatmap(pd.crosstab(titanic['Survived'], titanic['Pclass']))
```

```python
titanic.groupby('Pclass')['Survived'].mean()
```

## 6. Clustermap ( Categorical - Categorical)

```python
sns.clustermap(pd.crosstab(titanic['Parch'], titanic['Survived']))
```

## 7. Pairplot

```python
sns.pairplot(iris)
```

```python
sns.pairplot(data=iris, hue='Species')
```
